```python
import pandas as pd

file_path = "/content/es_combined_data.csv"

rows = []
bad_rows = []

with open(file_path, "r", encoding="utf-8", errors="replace") as f:

    # Read header
    header = next(f)
    print("Header:", header.strip())

    for line_num, line in enumerate(f, start=2):

        line = line.rstrip("\n\r")

        # Split ONLY from the right into 3 pieces
        parts = line.rsplit(",", 2)

        if len(parts) != 3:
            bad_rows.append((line_num, line))
            continue

        text, binary, multiclass = parts

        # Remove surrounding quotes from text if present
        if len(text) >= 2 and text.startswith('"') and text.endswith('"'):
            text = text[1:-1]

        # Standard CSV escaped quotes
        text = text.replace('""', '"')

        rows.append({
            "text": text,
            "binary": binary.strip(),
            "multiclass": multiclass.strip()
        })


df = pd.DataFrame(rows)

print("\nLoaded shape:", df.shape)
print("Bad rows:", len(bad_rows))

print("\nColumns:")
print(df.columns)

print("\nFirst 5 rows:")
print(df.head())
```

```python
import csv
import pandas as pd

file_path = "/content/es_combined_data.csv"

rows = []
bad_rows = []

with open(file_path, "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(
        f,
        quotechar='"',
        doublequote=True
    )

    header = next(reader)

    for record_number, row in enumerate(reader, start=2):
        if len(row) == 3:
            rows.append(row)
        else:
            bad_rows.append((record_number, row))

df = pd.DataFrame(rows, columns=header)

print("Header:", header)
print("Dataset shape:", df.shape)
print("Bad logical records:", len(bad_rows))
```

```python

# STEP 2: Check duplicate records

total_rows = len(df)

duplicate_extra = df.duplicated().sum()

all_duplicate_rows = df.duplicated(keep=False).sum()

unique_rows = len(df.drop_duplicates())

print("Total rows:", total_rows)
print("Duplicate extra copies:", duplicate_extra)
print("Rows involved in duplication:", all_duplicate_rows)
print("Unique rows after removing duplicates:", unique_rows)
```

```python
# STEP 3: Check repetition frequency of each unique record

record_counts = (
    df.groupby(["text", "binary", "multiclass"], dropna=False)
      .size()
)

print("How many unique records appear 1 time, 2 times, etc.:")
print(record_counts.value_counts().sort_index())

print("\nMaximum number of repetitions of any record:")
print(record_counts.max())
```

```python
import pandas as pd

es_train = pd.read_csv("/content/es_train.csv")
es_dev = pd.read_csv("/content/es_dev.csv")

print("Train shape:", es_train.shape)
print("Dev shape:", es_dev.shape)

print("\nTrain columns:", es_train.columns.tolist())
print("Dev columns:", es_dev.columns.tolist())
```

```python
# STEP 5: Create the 5,111 duplicated unique records

candidate_test = (
    df[df.duplicated(
        subset=["text", "binary", "multiclass"],
        keep=False
    )]
    .drop_duplicates(
        subset=["text", "binary", "multiclass"]
    )
    .reset_index(drop=True)
)

# Unique versions of train/dev for exact comparison
train_unique = es_train.drop_duplicates(
    subset=["text", "binary", "multiclass"]
)

dev_unique = es_dev.drop_duplicates(
    subset=["text", "binary", "multiclass"]
)

# Exact overlap on text + both labels
train_overlap = candidate_test.merge(
    train_unique,
    on=["text", "binary", "multiclass"],
    how="inner"
)

dev_overlap = candidate_test.merge(
    dev_unique,
    on=["text", "binary", "multiclass"],
    how="inner"
)

print("Candidate test size:", candidate_test.shape)
print("Exact overlap with train:", len(train_overlap))
print("Exact overlap with dev:", len(dev_overlap))
```

```python
# STEP 6: Check overlap based only on text

train_texts = set(es_train["text"].astype(str))
dev_texts = set(es_dev["text"].astype(str))
test_texts = set(candidate_test["text"].astype(str))

train_text_overlap = test_texts & train_texts
dev_text_overlap = test_texts & dev_texts

print("Text-only overlap with train:", len(train_text_overlap))
print("Text-only overlap with dev:", len(dev_text_overlap))
```

```python
# STEP 7: Save the recovered Spanish test set

es_test = candidate_test.copy()

es_test.to_csv(
    "/content/es_test.csv",
    index=False,
    encoding="utf-8"
)

print("Saved file: /content/es_test.csv")
print("Test shape:", es_test.shape)
```
