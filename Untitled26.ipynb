```python
import torch
from torch.utils.data import Dataset

MAX_LENGTH = 256

class HopeSpeechDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length):

        self.texts = dataframe["model_text"].tolist()
        self.labels = dataframe["label"].tolist()

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):
        return len(self.texts)


    def __getitem__(self, idx):

        text = self.texts[idx]
        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            truncation=True,
            padding=False
        )

        return {
            "input_ids": encoding["input_ids"],
            "attention_mask": encoding["attention_mask"],
            "labels": label
        }
```

```python
import numpy as np

def get_token_lengths(texts, tokenizer):
    lengths = []

    for text in texts:
        encoded = tokenizer(
            str(text),
            add_special_tokens=True,
            truncation=False,
            padding=False
        )

        lengths.append(len(encoded["input_ids"]))

    return np.array(lengths)


train_lengths = get_token_lengths(train_df["text"], tokenizer)
dev_lengths = get_token_lengths(dev_df["text"], tokenizer)
test_lengths = get_token_lengths(test_df["text"], tokenizer)
```

```python
def show_length_statistics(name, lengths):
    print(f"\n{name}")
    print("-" * 40)
    print("Total examples :", len(lengths))
    print("Minimum       :", np.min(lengths))
    print("Median (50%)  :", np.percentile(lengths, 50))
    print("90th percentile:", np.percentile(lengths, 90))
    print("95th percentile:", np.percentile(lengths, 95))
    print("99th percentile:", np.percentile(lengths, 99))
    print("Maximum       :", np.max(lengths))


show_length_statistics("TRAIN", train_lengths)
show_length_statistics("DEV", dev_lengths)
show_length_statistics("TEST", test_lengths)
```

```python
# ============================================
# STEP 5: PyTorch Dataset + Dynamic Padding
# ============================================

import torch
from torch.utils.data import Dataset, DataLoader
from transformers import DataCollatorWithPadding

# Maximum allowed sequence length
MAX_LENGTH = 256

# Batch size
BATCH_SIZE = 16


# --------------------------------------------
# 1. Create PyTorch Dataset Class
# --------------------------------------------

class HopeSpeechDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length):

        # Original raw tweets
        self.texts = dataframe["text"].astype(str).tolist()

        # Numeric labels created in previous step
        self.labels = dataframe["label"].astype(int).tolist()

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):
        return len(self.texts)


    def __getitem__(self, idx):

        text = self.texts[idx]
        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            add_special_tokens=True,
            max_length=self.max_length,
            truncation=True,
            padding=False
        )

        return {
            "input_ids": encoding["input_ids"],
            "attention_mask": encoding["attention_mask"],
            "labels": label
        }


# --------------------------------------------
# 2. Create Train, Dev and Test datasets
# --------------------------------------------

train_dataset = HopeSpeechDataset(
    train_df,
    tokenizer,
    MAX_LENGTH
)

dev_dataset = HopeSpeechDataset(
    dev_df,
    tokenizer,
    MAX_LENGTH
)

test_dataset = HopeSpeechDataset(
    test_df,
    tokenizer,
    MAX_LENGTH
)


print("Train samples:", len(train_dataset))
print("Dev samples:", len(dev_dataset))
print("Test samples:", len(test_dataset))


# --------------------------------------------
# 3. Dynamic Padding
# --------------------------------------------

data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer,
    padding=True,
    return_tensors="pt"
)


# --------------------------------------------
# 4. Create DataLoaders
# --------------------------------------------

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=data_collator
)

dev_loader = DataLoader(
    dev_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=data_collator
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=data_collator
)


# --------------------------------------------
# 5. Inspect one batch
# --------------------------------------------

batch = next(iter(train_loader))

print("\nInput IDs shape:")
print(batch["input_ids"].shape)

print("\nAttention Mask shape:")
print(batch["attention_mask"].shape)

print("\nLabels shape:")
print(batch["labels"].shape)

print("\nLabels:")
print(batch["labels"])


# --------------------------------------------
# 6. Check GPU
# --------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nUsing device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```
