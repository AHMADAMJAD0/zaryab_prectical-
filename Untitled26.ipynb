```python
train_df = pd.read_csv("/content/en_train.csv")
dev_df = pd.read_csv("/content/en_dev.csv")
test_df = pd.read_csv("/content/en_test_with_labels.csv")
```

```python
# Step 3: Label Encoding for Multiclass Classification

label2id = {
    "Not Hope": 0,
    "Generalized Hope": 1,
    "Realistic Hope": 2,
    "Unrealistic Hope": 3,
    "Sarcasm": 4
}

id2label = {
    0: "Not Hope",
    1: "Generalized Hope",
    2: "Realistic Hope",
    3: "Unrealistic Hope",
    4: "Sarcasm"
}

# Convert text labels into numeric labels
train_df["label"] = train_df["multiclass"].map(label2id)
dev_df["label"] = dev_df["multiclass"].map(label2id)
test_df["label"] = test_df["multiclass"].map(label2id)

# Check whether any unknown label exists
print("Missing encoded labels:")
print("Train:", train_df["label"].isnull().sum())
print("Dev:", dev_df["label"].isnull().sum())
print("Test:", test_df["label"].isnull().sum())

# Show sample
print("\nTraining sample:")
print(train_df[["text", "multiclass", "label"]].head())

# Check distribution
print("\nTrain label distribution:")
print(train_df["label"].value_counts().sort_index())

print("\nDev label distribution:")
print(dev_df["label"].value_counts().sort_index())

print("\nTest label distribution:")
print(test_df["label"].value_counts().sort_index())
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "FacebookAI/roberta-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer loaded successfully!")
print("Model:", MODEL_NAME)
print("Vocabulary size:", tokenizer.vocab_size)
print("Maximum model length:", tokenizer.model_max_length)
```

```python
sample_text = train_df["text"].iloc[0]

print("Original text:")
print(sample_text)

tokens = tokenizer.tokenize(sample_text)

print("\nTokens:")
print(tokens)

encoded = tokenizer(
    sample_text,
    add_special_tokens=True,
    return_attention_mask=True
)

print("\nInput IDs:")
print(encoded["input_ids"])

print("\nAttention Mask:")
print(encoded["attention_mask"])

print("\nNumber of tokens:")
print(len(encoded["input_ids"]))
```

```python
import numpy as np

def get_token_lengths(texts, tokenizer):
    lengths = []

    for text in texts:
        encoded = tokenizer(
            str(text),
            add_special_tokens=True,
            truncation=False,
            padding=False
        )

        lengths.append(len(encoded["input_ids"]))

    return np.array(lengths)


train_lengths = get_token_lengths(train_df["text"], tokenizer)
dev_lengths = get_token_lengths(dev_df["text"], tokenizer)
test_lengths = get_token_lengths(test_df["text"], tokenizer)
```

```python
def show_length_statistics(name, lengths):
    print(f"\n{name}")
    print("-" * 40)
    print("Total examples :", len(lengths))
    print("Minimum       :", np.min(lengths))
    print("Median (50%)  :", np.percentile(lengths, 50))
    print("90th percentile:", np.percentile(lengths, 90))
    print("95th percentile:", np.percentile(lengths, 95))
    print("99th percentile:", np.percentile(lengths, 99))
    print("Maximum       :", np.max(lengths))


show_length_statistics("TRAIN", train_lengths)
show_length_statistics("DEV", dev_lengths)
show_length_statistics("TEST", test_lengths)
```

```python
for max_len in [64, 128, 256, 512]:
    train_over = np.sum(train_lengths > max_len)
    train_percent = (train_over / len(train_lengths)) * 100

    print(
        f"More than {max_len:3d} tokens: "
        f"{train_over:4d} examples "
        f"({train_percent:.2f}%)"
    )
```
