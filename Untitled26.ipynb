```python
import torch
from torch.utils.data import Dataset

MAX_LENGTH = 256

class HopeSpeechDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length):

        self.texts = dataframe["model_text"].tolist()
        self.labels = dataframe["label"].tolist()

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):
        return len(self.texts)


    def __getitem__(self, idx):

        text = self.texts[idx]
        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            truncation=True,
            padding=False
        )

        return {
            "input_ids": encoding["input_ids"],
            "attention_mask": encoding["attention_mask"],
            "labels": label
        }
```

```python
import numpy as np
import torch

from sklearn.metrics import (
    f1_score,
    accuracy_score
)


def evaluate_model(model, dataloader, device):

    # Evaluation mode
    model.eval()

    all_predictions = []
    all_labels = []

    # No gradient calculation during evaluation
    with torch.no_grad():

        for batch in dataloader:

            # Move batch to GPU
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].to(device)

            # Forward pass
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            # Raw model outputs for 5 classes
            logits = outputs.logits

            # Select class with highest score
            predictions = torch.argmax(
                logits,
                dim=1
            )

            # Store results on CPU
            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )


    # Calculate metrics
    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        average="macro"
    )

    weighted_f1 = f1_score(
        all_labels,
        all_predictions,
        average="weighted"
    )

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )


    return {
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1,
        "accuracy": accuracy
    }
```

```python
def train_one_epoch(model, dataloader, optimizer, scheduler, device):

    # Put model in training mode
    model.train()

    total_loss = 0.0

    # Go through all batches
    for batch in dataloader:

        # Move batch to GPU
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        # Clear gradients from previous batch
        optimizer.zero_grad()

        # Forward pass
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )

        # Cross-entropy loss
        loss = outputs.loss

        # Backpropagation
        loss.backward()

        # Prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        # Update model weights
        optimizer.step()

        # Update learning-rate scheduler
        scheduler.step()

        # Save batch loss
        total_loss += loss.item()

    # Average loss across all batches
    average_loss = total_loss / len(dataloader)

    return average_loss
```

```python
id2label = {
    0: "Not Hope",
    1: "Generalized Hope",
    2: "Realistic Hope",
    3: "Unrealistic Hope",
    4: "Sarcasm"
}
```

```python
import gc
import torch
import optuna

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup
)

MODEL_NAME = "cardiffnlp/twitter-roberta-base"

def objective(trial):

    # -----------------------------------------
    # 1. Optuna chooses hyperparameters
    # -----------------------------------------

    learning_rate = trial.suggest_float(
        "learning_rate",
        1e-5,
        5e-5,
        log=True
    )

    batch_size = trial.suggest_categorical(
        "batch_size",
        [8, 16]
    )

    weight_decay = trial.suggest_float(
        "weight_decay",
        0.0,
        0.1
    )

    num_epochs = trial.suggest_int(
        "num_epochs",
        2,
        5
    )

    warmup_ratio = trial.suggest_float(
        "warmup_ratio",
        0.0,
        0.15
    )


    # -----------------------------------------
    # 2. Create DataLoaders for this trial
    # -----------------------------------------

    train_loader_trial = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=data_collator
    )

    dev_loader_trial = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=data_collator
    )


    # -----------------------------------------
    # 3. Load a FRESH model
    # -----------------------------------------

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=5,
        label2id=label2id,
        id2label=id2label
    )

    model.to(device)


    # -----------------------------------------
    # 4. Optimizer
    # -----------------------------------------

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )


    # -----------------------------------------
    # 5. Scheduler
    # -----------------------------------------

    total_training_steps = (
        len(train_loader_trial) * num_epochs
    )

    warmup_steps = int(
        total_training_steps * warmup_ratio
    )

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_training_steps
    )


    best_macro_f1 = 0.0


    # -----------------------------------------
    # 6. Training epochs
    # -----------------------------------------

    for epoch in range(num_epochs):

        train_loss = train_one_epoch(
            model=model,
            dataloader=train_loader_trial,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device
        )

        dev_results = evaluate_model(
            model=model,
            dataloader=dev_loader_trial,
            device=device
        )

        macro_f1 = dev_results["macro_f1"]

        print(
            f"Trial {trial.number} | "
            f"Epoch {epoch + 1}/{num_epochs} | "
            f"Loss: {train_loss:.4f} | "
            f"Dev Macro-F1: {macro_f1:.4f}"
        )


        # Keep best DEV score from this trial
        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1


        # Tell Optuna how trial is progressing
        trial.report(
            macro_f1,
            step=epoch
        )


        # Stop bad trials early
        if trial.should_prune():

            del model
            torch.cuda.empty_cache()
            gc.collect()

            raise optuna.TrialPruned()


    # -----------------------------------------
    # 7. Clean GPU memory
    # -----------------------------------------

    del model
    torch.cuda.empty_cache()
    gc.collect()


    # Optuna will maximize this value
    return best_macro_f1
```

```python
import gc
import torch
import optuna

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup,
    set_seed
)

MODEL_NAME = "cardiffnlp/twitter-roberta-base"

MAX_EPOCHS = 5
EARLY_STOPPING_PATIENCE = 2
SEED = 42


def objective(trial):

    # --------------------------------------------------
    # 1. Reproducibility
    # --------------------------------------------------
    set_seed(SEED)


    # --------------------------------------------------
    # 2. Optuna chooses hyperparameters
    # --------------------------------------------------

    learning_rate = trial.suggest_float(
        "learning_rate",
        1e-5,
        5e-5,
        log=True
    )

    batch_size = trial.suggest_categorical(
        "batch_size",
        [8, 16]
    )

    weight_decay = trial.suggest_float(
        "weight_decay",
        0.0,
        0.1
    )

    warmup_ratio = trial.suggest_float(
        "warmup_ratio",
        0.0,
        0.15
    )


    # --------------------------------------------------
    # 3. DataLoaders for this trial
    # --------------------------------------------------

    train_loader_trial = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=data_collator
    )

    dev_loader_trial = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=data_collator
    )


    # --------------------------------------------------
    # 4. Fresh model for every trial
    # --------------------------------------------------

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=5,
        label2id=label2id,
        id2label=id2label
    )

    model.to(device)


    # --------------------------------------------------
    # 5. Optimizer
    # --------------------------------------------------

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )


    # --------------------------------------------------
    # 6. Scheduler
    # --------------------------------------------------

    total_training_steps = (
        len(train_loader_trial) * MAX_EPOCHS
    )

    warmup_steps = int(
        total_training_steps * warmup_ratio
    )

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_training_steps
    )


    # --------------------------------------------------
    # 7. Training
    # --------------------------------------------------

    best_macro_f1 = 0.0
    best_epoch = 0
    epochs_without_improvement = 0


    for epoch in range(MAX_EPOCHS):

        train_loss = train_one_epoch(
            model=model,
            dataloader=train_loader_trial,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device
        )


        dev_results = evaluate_model(
            model=model,
            dataloader=dev_loader_trial,
            device=device
        )


        macro_f1 = dev_results["macro_f1"]


        print(
            f"Trial {trial.number} | "
            f"Epoch {epoch + 1}/{MAX_EPOCHS} | "
            f"Loss: {train_loss:.4f} | "
            f"Dev Macro-F1: {macro_f1:.4f}"
        )


        # --------------------------------------------------
        # 8. Track best epoch
        # --------------------------------------------------

        if macro_f1 > best_macro_f1:

            best_macro_f1 = macro_f1
            best_epoch = epoch + 1

            epochs_without_improvement = 0

        else:

            epochs_without_improvement += 1


        # Tell Optuna current result
        trial.report(
            macro_f1,
            step=epoch
        )


        # --------------------------------------------------
        # 9. Optuna pruning
        # --------------------------------------------------

        if trial.should_prune():

            del model
            torch.cuda.empty_cache()
            gc.collect()

            raise optuna.TrialPruned()


        # --------------------------------------------------
        # 10. Early stopping
        # --------------------------------------------------

        if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

            print(
                f"Early stopping at epoch {epoch + 1}. "
                f"Best epoch: {best_epoch}"
            )

            break


    # Store best epoch for later inspection
    trial.set_user_attr(
        "best_epoch",
        best_epoch
    )


    # --------------------------------------------------
    # 11. Clean GPU memory
    # --------------------------------------------------

    del model
    torch.cuda.empty_cache()
    gc.collect()


    return best_macro_f1
```

```python
sampler = optuna.samplers.TPESampler(
    seed=42
)

pruner = optuna.pruners.MedianPruner(
    n_startup_trials=3,
    n_warmup_steps=1
)

final_study = optuna.create_study(
    direction="maximize",
    sampler=sampler,
    pruner=pruner
)

final_study.optimize(
    objective,
    n_trials=15
)

print("\nBest DEV Macro-F1:")
print(final_study.best_value)

print("\nBest hyperparameters:")
print(final_study.best_params)

print("\nBest epoch:")
print(
    final_study.best_trial.user_attrs["best_epoch"]
)
```
