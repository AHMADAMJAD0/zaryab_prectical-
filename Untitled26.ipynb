```python
import torch
from torch.utils.data import Dataset

MAX_LENGTH = 256

class HopeSpeechDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length):

        self.texts = dataframe["model_text"].tolist()
        self.labels = dataframe["label"].tolist()

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):
        return len(self.texts)


    def __getitem__(self, idx):

        text = self.texts[idx]
        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            truncation=True,
            padding=False
        )

        return {
            "input_ids": encoding["input_ids"],
            "attention_mask": encoding["attention_mask"],
            "labels": label
        }
```

```python
import numpy as np
import torch

from sklearn.metrics import (
    f1_score,
    accuracy_score
)


def evaluate_model(model, dataloader, device):

    # Evaluation mode
    model.eval()

    all_predictions = []
    all_labels = []

    # No gradient calculation during evaluation
    with torch.no_grad():

        for batch in dataloader:

            # Move batch to GPU
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].to(device)

            # Forward pass
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            # Raw model outputs for 5 classes
            logits = outputs.logits

            # Select class with highest score
            predictions = torch.argmax(
                logits,
                dim=1
            )

            # Store results on CPU
            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )


    # Calculate metrics
    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        average="macro"
    )

    weighted_f1 = f1_score(
        all_labels,
        all_predictions,
        average="weighted"
    )

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )


    return {
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1,
        "accuracy": accuracy
    }
```

```python
def train_one_epoch(model, dataloader, optimizer, scheduler, device):

    # Put model in training mode
    model.train()

    total_loss = 0.0

    # Go through all batches
    for batch in dataloader:

        # Move batch to GPU
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        # Clear gradients from previous batch
        optimizer.zero_grad()

        # Forward pass
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )

        # Cross-entropy loss
        loss = outputs.loss

        # Backpropagation
        loss.backward()

        # Prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        # Update model weights
        optimizer.step()

        # Update learning-rate scheduler
        scheduler.step()

        # Save batch loss
        total_loss += loss.item()

    # Average loss across all batches
    average_loss = total_loss / len(dataloader)

    return average_loss
```

```python
import gc
import torch
import optuna

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup
)

MODEL_NAME = "cardiffnlp/twitter-roberta-base"

def objective(trial):

    # -----------------------------------------
    # 1. Optuna chooses hyperparameters
    # -----------------------------------------

    learning_rate = trial.suggest_float(
        "learning_rate",
        1e-5,
        5e-5,
        log=True
    )

    batch_size = trial.suggest_categorical(
        "batch_size",
        [8, 16]
    )

    weight_decay = trial.suggest_float(
        "weight_decay",
        0.0,
        0.1
    )

    num_epochs = trial.suggest_int(
        "num_epochs",
        2,
        5
    )

    warmup_ratio = trial.suggest_float(
        "warmup_ratio",
        0.0,
        0.15
    )


    # -----------------------------------------
    # 2. Create DataLoaders for this trial
    # -----------------------------------------

    train_loader_trial = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=data_collator
    )

    dev_loader_trial = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=data_collator
    )


    # -----------------------------------------
    # 3. Load a FRESH model
    # -----------------------------------------

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=5,
        label2id=label2id,
        id2label=id2label
    )

    model.to(device)


    # -----------------------------------------
    # 4. Optimizer
    # -----------------------------------------

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )


    # -----------------------------------------
    # 5. Scheduler
    # -----------------------------------------

    total_training_steps = (
        len(train_loader_trial) * num_epochs
    )

    warmup_steps = int(
        total_training_steps * warmup_ratio
    )

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_training_steps
    )


    best_macro_f1 = 0.0


    # -----------------------------------------
    # 6. Training epochs
    # -----------------------------------------

    for epoch in range(num_epochs):

        train_loss = train_one_epoch(
            model=model,
            dataloader=train_loader_trial,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device
        )

        dev_results = evaluate_model(
            model=model,
            dataloader=dev_loader_trial,
            device=device
        )

        macro_f1 = dev_results["macro_f1"]

        print(
            f"Trial {trial.number} | "
            f"Epoch {epoch + 1}/{num_epochs} | "
            f"Loss: {train_loss:.4f} | "
            f"Dev Macro-F1: {macro_f1:.4f}"
        )


        # Keep best DEV score from this trial
        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1


        # Tell Optuna how trial is progressing
        trial.report(
            macro_f1,
            step=epoch
        )


        # Stop bad trials early
        if trial.should_prune():

            del model
            torch.cuda.empty_cache()
            gc.collect()

            raise optuna.TrialPruned()


    # -----------------------------------------
    # 7. Clean GPU memory
    # -----------------------------------------

    del model
    torch.cuda.empty_cache()
    gc.collect()


    # Optuna will maximize this value
    return best_macro_f1
```

```python
import gc
import torch
import optuna

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup,
    set_seed
)

MODEL_NAME = "cardiffnlp/twitter-roberta-base"

MAX_EPOCHS = 5
EARLY_STOPPING_PATIENCE = 2
SEED = 42


def objective(trial):

    # --------------------------------------------------
    # 1. Reproducibility
    # --------------------------------------------------
    set_seed(SEED)


    # --------------------------------------------------
    # 2. Optuna chooses hyperparameters
    # --------------------------------------------------

    learning_rate = trial.suggest_float(
        "learning_rate",
        1e-5,
        5e-5,
        log=True
    )

    batch_size = trial.suggest_categorical(
        "batch_size",
        [8, 16]
    )

    weight_decay = trial.suggest_float(
        "weight_decay",
        0.0,
        0.1
    )

    warmup_ratio = trial.suggest_float(
        "warmup_ratio",
        0.0,
        0.15
    )


    # --------------------------------------------------
    # 3. DataLoaders for this trial
    # --------------------------------------------------

    train_loader_trial = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=data_collator
    )

    dev_loader_trial = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=data_collator
    )


    # --------------------------------------------------
    # 4. Fresh model for every trial
    # --------------------------------------------------

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=5,
        label2id=label2id,
        id2label=id2label
    )

    model.to(device)


    # --------------------------------------------------
    # 5. Optimizer
    # --------------------------------------------------

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )


    # --------------------------------------------------
    # 6. Scheduler
    # --------------------------------------------------

    total_training_steps = (
        len(train_loader_trial) * MAX_EPOCHS
    )

    warmup_steps = int(
        total_training_steps * warmup_ratio
    )

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_training_steps
    )


    # --------------------------------------------------
    # 7. Training
    # --------------------------------------------------

    best_macro_f1 = 0.0
    best_epoch = 0
    epochs_without_improvement = 0


    for epoch in range(MAX_EPOCHS):

        train_loss = train_one_epoch(
            model=model,
            dataloader=train_loader_trial,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device
        )


        dev_results = evaluate_model(
            model=model,
            dataloader=dev_loader_trial,
            device=device
        )


        macro_f1 = dev_results["macro_f1"]


        print(
            f"Trial {trial.number} | "
            f"Epoch {epoch + 1}/{MAX_EPOCHS} | "
            f"Loss: {train_loss:.4f} | "
            f"Dev Macro-F1: {macro_f1:.4f}"
        )


        # --------------------------------------------------
        # 8. Track best epoch
        # --------------------------------------------------

        if macro_f1 > best_macro_f1:

            best_macro_f1 = macro_f1
            best_epoch = epoch + 1

            epochs_without_improvement = 0

        else:

            epochs_without_improvement += 1


        # Tell Optuna current result
        trial.report(
            macro_f1,
            step=epoch
        )


        # --------------------------------------------------
        # 9. Optuna pruning
        # --------------------------------------------------

        if trial.should_prune():

            del model
            torch.cuda.empty_cache()
            gc.collect()

            raise optuna.TrialPruned()


        # --------------------------------------------------
        # 10. Early stopping
        # --------------------------------------------------

        if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

            print(
                f"Early stopping at epoch {epoch + 1}. "
                f"Best epoch: {best_epoch}"
            )

            break


    # Store best epoch for later inspection
    trial.set_user_attr(
        "best_epoch",
        best_epoch
    )


    # --------------------------------------------------
    # 11. Clean GPU memory
    # --------------------------------------------------

    del model
    torch.cuda.empty_cache()
    gc.collect()


    return best_macro_f1
```

```python
import copy
import torch

from torch.utils.data import DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup,
    set_seed
)

# ============================================================
# FINAL SETTINGS FROM OPTUNA
# ============================================================

MODEL_NAME = "cardiffnlp/twitter-roberta-base"

BEST_LR = 3.538461259525519e-05
BEST_BATCH_SIZE = 16
BEST_WEIGHT_DECAY = 0.05924145688620425
BEST_WARMUP_RATIO = 0.006967561907999659

# Optuna found epoch 3 to be best
FINAL_EPOCHS = 3

SEED = 42
set_seed(SEED)


# ============================================================
# DATALOADERS
# ============================================================

train_loader_final = DataLoader(
    train_dataset,
    batch_size=BEST_BATCH_SIZE,
    shuffle=True,
    collate_fn=data_collator
)

dev_loader_final = DataLoader(
    dev_dataset,
    batch_size=BEST_BATCH_SIZE,
    shuffle=False,
    collate_fn=data_collator
)

test_loader_final = DataLoader(
    test_dataset,
    batch_size=BEST_BATCH_SIZE,
    shuffle=False,
    collate_fn=data_collator
)


# ============================================================
# FRESH MODEL
# ============================================================

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=5,
    label2id=label2id,
    id2label=id2label,
    problem_type="single_label_classification"
)

model.to(device)


# ============================================================
# OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=BEST_LR,
    weight_decay=BEST_WEIGHT_DECAY
)


# ============================================================
# SCHEDULER
# ============================================================

total_training_steps = (
    len(train_loader_final) * FINAL_EPOCHS
)

warmup_steps = int(
    total_training_steps * BEST_WARMUP_RATIO
)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_training_steps
)


# ============================================================
# FINAL TRAINING
# ============================================================

best_dev_f1 = 0.0
best_model_state = None
best_epoch = 0

for epoch in range(FINAL_EPOCHS):

    train_loss = train_one_epoch(
        model=model,
        dataloader=train_loader_final,
        optimizer=optimizer,
        scheduler=scheduler,
        device=device
    )

    dev_results = evaluate_model(
        model=model,
        dataloader=dev_loader_final,
        device=device
    )

    dev_macro_f1 = dev_results["macro_f1"]

    print(
        f"Epoch {epoch + 1}/{FINAL_EPOCHS} | "
        f"Loss: {train_loss:.4f} | "
        f"Dev Macro-F1: {dev_macro_f1:.4f}"
    )

    # Save best DEV checkpoint
    if dev_macro_f1 > best_dev_f1:

        best_dev_f1 = dev_macro_f1
        best_epoch = epoch + 1

        best_model_state = copy.deepcopy(
            model.state_dict()
        )


print("\nBest DEV Macro-F1:", best_dev_f1)
print("Best epoch:", best_epoch)


# ============================================================
# RESTORE BEST MODEL
# ============================================================

model.load_state_dict(best_model_state)

print("\nBest model checkpoint restored.")
```

```python
# ============================================================
# ALL ENGLISH MODELS + OPTUNA
# Maximum Epochs = 25
# ============================================================

import gc
import copy
import os
import numpy as np
import pandas as pd
import torch
import optuna

from torch.utils.data import Dataset, DataLoader

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    get_linear_schedule_with_warmup,
    set_seed
)

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report
)


# ============================================================
# 1. GLOBAL SETTINGS
# ============================================================

SEED = 42

MAX_LENGTH = 256

MAX_EPOCHS = 25

EARLY_STOPPING_PATIENCE = 2

N_TRIALS = 15

NUM_LABELS = 5

GRADIENT_CLIP = 1.0


device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# ============================================================
# 2. LABELS
# ============================================================

label2id = {
    "Not Hope": 0,
    "Generalized Hope": 1,
    "Realistic Hope": 2,
    "Unrealistic Hope": 3,
    "Sarcasm": 4
}

id2label = {
    0: "Not Hope",
    1: "Generalized Hope",
    2: "Realistic Hope",
    3: "Unrealistic Hope",
    4: "Sarcasm"
}


# ============================================================
# 3. MODELS TO TEST
# ============================================================

MODELS = {

    "E0_General_RoBERTa":
        "FacebookAI/roberta-base",

    "E1_Twitter_RoBERTa":
        "cardiffnlp/twitter-roberta-base",

    "E2_Twitter_RoBERTa_Emotion":
        "cardiffnlp/twitter-roberta-base-emotion",

    "E3_Twitter_RoBERTa_Emotion_Latest":
        "cardiffnlp/twitter-roberta-base-emotion-latest"
}


# ============================================================
# 4. PYTORCH DATASET
# ============================================================

class HopeSpeechDataset(Dataset):

    def __init__(
        self,
        dataframe,
        tokenizer,
        max_length
    ):

        self.texts = (
            dataframe["model_text"]
            .astype(str)
            .tolist()
        )

        self.labels = (
            dataframe["label"]
            .astype(int)
            .tolist()
        )

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):

        return len(self.texts)


    def __getitem__(self, idx):

        text = self.texts[idx]

        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            max_length=self.max_length,
            truncation=True,
            padding=False
        )

        return {

            "input_ids":
                encoding["input_ids"],

            "attention_mask":
                encoding["attention_mask"],

            "labels":
                label
        }


# ============================================================
# 5. EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model,
    dataloader,
    device
):

    model.eval()

    predictions = []
    true_labels = []

    with torch.no_grad():

        for batch in dataloader:

            input_ids = (
                batch["input_ids"]
                .to(device)
            )

            attention_mask = (
                batch["attention_mask"]
                .to(device)
            )

            labels = (
                batch["labels"]
                .to(device)
            )

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

            preds = torch.argmax(
                logits,
                dim=1
            )

            predictions.extend(
                preds.cpu().numpy()
            )

            true_labels.extend(
                labels.cpu().numpy()
            )


    macro_f1 = f1_score(
        true_labels,
        predictions,
        average="macro"
    )

    weighted_f1 = f1_score(
        true_labels,
        predictions,
        average="weighted"
    )

    accuracy = accuracy_score(
        true_labels,
        predictions
    )


    return {

        "macro_f1":
            macro_f1,

        "weighted_f1":
            weighted_f1,

        "accuracy":
            accuracy,

        "predictions":
            np.array(predictions),

        "labels":
            np.array(true_labels)
    }


# ============================================================
# 6. TRAIN ONE EPOCH
# ============================================================

def train_one_epoch(
    model,
    dataloader,
    optimizer,
    scheduler,
    device
):

    model.train()

    total_loss = 0.0


    for batch in dataloader:

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )


        optimizer.zero_grad()


        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )


        loss = outputs.loss


        loss.backward()


        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            GRADIENT_CLIP
        )


        optimizer.step()

        scheduler.step()


        total_loss += loss.item()


    average_loss = (
        total_loss /
        len(dataloader)
    )


    return average_loss


# ============================================================
# 7. LOAD FRESH MODEL
# ============================================================

def load_fresh_model(
    model_name
):

    model = (
        AutoModelForSequenceClassification
        .from_pretrained(

            model_name,

            num_labels=NUM_LABELS,

            label2id=label2id,

            id2label=id2label,

            problem_type=
            "single_label_classification",

            # Needed especially for emotion models
            # whose old classifier may have
            # 4/11 emotion labels instead of 5 Hope labels.
            ignore_mismatched_sizes=True
        )
    )


    model.to(device)

    return model


# ============================================================
# 8. PREPARE DATA FOR EACH MODEL
# ============================================================

def prepare_model_data(
    model_name
):

    print("\nLoading tokenizer:")
    print(model_name)


    tokenizer = (
        AutoTokenizer
        .from_pretrained(
            model_name
        )
    )


    train_dataset = HopeSpeechDataset(
        train_df,
        tokenizer,
        MAX_LENGTH
    )


    dev_dataset = HopeSpeechDataset(
        dev_df,
        tokenizer,
        MAX_LENGTH
    )


    test_dataset = HopeSpeechDataset(
        test_df,
        tokenizer,
        MAX_LENGTH
    )


    data_collator = (
        DataCollatorWithPadding(
            tokenizer=tokenizer,
            padding=True,
            return_tensors="pt"
        )
    )


    return (
        tokenizer,
        train_dataset,
        dev_dataset,
        test_dataset,
        data_collator
    )


# ============================================================
# 9. CREATE OPTUNA OBJECTIVE
# ============================================================

def create_objective(
    model_name,
    train_dataset,
    dev_dataset,
    data_collator
):

    def objective(trial):

        set_seed(SEED)


        # ====================================================
        # OPTUNA PARAMETERS
        # ====================================================

        learning_rate = (
            trial.suggest_float(
                "learning_rate",
                1e-5,
                5e-5,
                log=True
            )
        )


        batch_size = (
            trial.suggest_categorical(
                "batch_size",
                [8, 16]
            )
        )


        weight_decay = (
            trial.suggest_float(
                "weight_decay",
                0.0,
                0.1
            )
        )


        warmup_ratio = (
            trial.suggest_float(
                "warmup_ratio",
                0.0,
                0.15
            )
        )


        # Optuna may choose 3–25 epochs.
        # 25 is the maximum.
        num_epochs = (
            trial.suggest_int(
                "num_epochs",
                3,
                MAX_EPOCHS
            )
        )


        # ====================================================
        # DATALOADERS
        # ====================================================

        train_loader = DataLoader(

            train_dataset,

            batch_size=batch_size,

            shuffle=True,

            collate_fn=data_collator
        )


        dev_loader = DataLoader(

            dev_dataset,

            batch_size=batch_size,

            shuffle=False,

            collate_fn=data_collator
        )


        # ====================================================
        # FRESH MODEL FOR EVERY TRIAL
        # ====================================================

        model = load_fresh_model(
            model_name
        )


        # ====================================================
        # OPTIMIZER
        # ====================================================

        optimizer = torch.optim.AdamW(

            model.parameters(),

            lr=learning_rate,

            weight_decay=weight_decay
        )


        # ====================================================
        # LR SCHEDULER
        # ====================================================

        total_steps = (

            len(train_loader)
            *
            num_epochs
        )


        warmup_steps = int(

            total_steps
            *
            warmup_ratio
        )


        scheduler = (
            get_linear_schedule_with_warmup(

                optimizer,

                num_warmup_steps=
                    warmup_steps,

                num_training_steps=
                    total_steps
            )
        )


        # ====================================================
        # TRAINING
        # ====================================================

        best_macro_f1 = 0.0

        best_epoch = 0

        no_improvement = 0


        for epoch in range(
            num_epochs
        ):

            train_loss = (
                train_one_epoch(

                    model,

                    train_loader,

                    optimizer,

                    scheduler,

                    device
                )
            )


            dev_results = (
                evaluate_model(

                    model,

                    dev_loader,

                    device
                )
            )


            current_f1 = (
                dev_results[
                    "macro_f1"
                ]
            )


            print(

                f"Trial {trial.number} | "

                f"Epoch "
                f"{epoch + 1}/"
                f"{num_epochs} | "

                f"Loss: "
                f"{train_loss:.4f} | "

                f"Dev Macro-F1: "
                f"{current_f1:.4f}"
            )


            # =================================================
            # BEST EPOCH
            # =================================================

            if (
                current_f1
                >
                best_macro_f1
            ):

                best_macro_f1 = (
                    current_f1
                )

                best_epoch = (
                    epoch + 1
                )

                no_improvement = 0

            else:

                no_improvement += 1


            # =================================================
            # REPORT TO OPTUNA
            # =================================================

            trial.report(
                current_f1,
                epoch
            )


            # =================================================
            # OPTUNA PRUNING
            # =================================================

            if trial.should_prune():

                del model

                torch.cuda.empty_cache()

                gc.collect()

                raise (
                    optuna
                    .TrialPruned()
                )


            # =================================================
            # EARLY STOPPING
            # =================================================

            if (
                no_improvement
                >=
                EARLY_STOPPING_PATIENCE
            ):

                print(
                    "Early stopping "
                    f"at epoch "
                    f"{epoch + 1}. "
                    f"Best epoch: "
                    f"{best_epoch}"
                )

                break


        # Save useful information
        trial.set_user_attr(
            "best_epoch",
            best_epoch
        )


        # ====================================================
        # CLEAN GPU MEMORY
        # ====================================================

        del model

        torch.cuda.empty_cache()

        gc.collect()


        return best_macro_f1


    return objective


# ============================================================
# 10. OPTUNA SEARCH FOR ONE MODEL
# ============================================================

def run_optuna_search(
    experiment_name,
    model_name
):

    print("\n")
    print("=" * 80)

    print(
        "EXPERIMENT:",
        experiment_name
    )

    print(
        "MODEL:",
        model_name
    )

    print("=" * 80)


    (
        tokenizer,
        train_dataset,
        dev_dataset,
        test_dataset,
        data_collator
    ) = prepare_model_data(
        model_name
    )


    objective = create_objective(

        model_name,

        train_dataset,

        dev_dataset,

        data_collator
    )


    # Same sampler for all models
    sampler = (
        optuna.samplers
        .TPESampler(
            seed=SEED
        )
    )


    # Same pruning strategy
    pruner = (
        optuna.pruners
        .MedianPruner(

            n_startup_trials=3,

            n_warmup_steps=1
        )
    )


    study = optuna.create_study(

        direction="maximize",

        sampler=sampler,

        pruner=pruner,

        study_name=
            experiment_name
    )


    study.optimize(

        objective,

        n_trials=N_TRIALS
    )


    print("\n")
    print(
        "BEST MODEL SETTINGS"
    )

    print(
        "Best Dev Macro-F1:",
        study.best_value
    )

    print(
        "Best parameters:",
        study.best_params
    )

    print(
        "Best epoch:",
        study.best_trial
             .user_attrs[
                 "best_epoch"
             ]
    )


    return {

        "study":
            study,

        "tokenizer":
            tokenizer,

        "train_dataset":
            train_dataset,

        "dev_dataset":
            dev_dataset,

        "test_dataset":
            test_dataset,

        "data_collator":
            data_collator
    }


# ============================================================
# 11. RUN OPTUNA FOR ALL FOUR MODELS
# ============================================================

ALL_RESULTS = {}


for (
    experiment_name,
    model_name
) in MODELS.items():


    result = run_optuna_search(

        experiment_name,

        model_name
    )


    ALL_RESULTS[
        experiment_name
    ] = result


# ============================================================
# 12. CREATE SUMMARY TABLE
# ============================================================

summary_rows = []


for (
    experiment_name,
    model_name
) in MODELS.items():


    study = (
        ALL_RESULTS[
            experiment_name
        ]["study"]
    )


    summary_rows.append({

        "Experiment":
            experiment_name,

        "Model":
            model_name,

        "Best_DEV_Macro_F1":
            study.best_value,

        "Best_Epoch":
            study.best_trial
                 .user_attrs[
                     "best_epoch"
                 ],

        "Planned_Epochs":
            study.best_params[
                "num_epochs"
            ],

        "Learning_Rate":
            study.best_params[
                "learning_rate"
            ],

        "Batch_Size":
            study.best_params[
                "batch_size"
            ],

        "Weight_Decay":
            study.best_params[
                "weight_decay"
            ],

        "Warmup_Ratio":
            study.best_params[
                "warmup_ratio"
            ]
    })


summary_df = pd.DataFrame(
    summary_rows
)


print("\n")
print("=" * 80)
print("FINAL OPTUNA SUMMARY")
print("=" * 80)

print(summary_df)


# Save summary
summary_df.to_csv(
    "/content/english_optuna_summary.csv",
    index=False
)

print(
    "\nSaved:"
    "/content/"
    "english_optuna_summary.csv"
)
```
